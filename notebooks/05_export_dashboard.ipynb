# 05 · Exportar resultados al dashboard

Reúne los resultados de los notebooks 00–04 (solo agregados; ningún identificador ni fila individual) en
`dashboard/data/dashboard_data.json` y los incrusta en `dashboard/index.html` para abrirlo con doble clic.

```python
import json
import runpy
import config

datos = {
    "meta": {"empresa": "una empresa de telefonía móvil en Colombia", "periodo": "enero–octubre de 2026",
             "unidad": "registro de línea móvil"},
    "preparacion": config.leer_json("r00_preparacion"),
    "descriptiva": config.leer_json("r01_descriptiva"),
    "regresion": config.leer_json("r02_regresion"),
    "probabilidad": config.leer_json("r03_probabilidad"),
    "ciencia": config.leer_json("r04_ciencia_datos"),
}
config.EXPORT.mkdir(parents=True, exist_ok=True)
ruta = config.EXPORT / "dashboard_data.json"
ruta.write_text(json.dumps(datos, ensure_ascii=False, allow_nan=False), encoding="utf-8")
print(f"Exportado: {ruta.name} ({ruta.stat().st_size / 1024:.0f} KB)")
```

```python
_ = runpy.run_path(str(config.DASHBOARD / "build.py"))
```
